# Load Dataset

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.remote("sc://localhost").getOrCreate()
spark
```

```python
df = spark.read.csv("/home/analytics4220/Taylor/warehouse_sales.csv",
                   header= True, quote = '"')
```

```python
df.show()
```

## Change number columns (previously strings) to float

```python
from pyspark.sql.functions import when

df = df.withColumn(
    "WAREHOUSE SALES",
    when(
        df["WAREHOUSE SALES"].rlike(r'^-?[0-9]+(\.[0-9]+)?$'),
        df["WAREHOUSE SALES"].cast("float")
    ).otherwise(None)
)

df = df.withColumn(
    "RETAIL SALES",
    when(
        df["RETAIL SALES"].rlike(r'^-?[0-9]+(\.[0-9]+)?$'),
        df["RETAIL SALES"].cast("float")
    ).otherwise(None)
)

df = df.withColumn(
    "RETAIL TRANSFERS",
    when(
        df["RETAIL TRANSFERS"].rlike(r'^-?[0-9]+(\.[0-9]+)?$'),
        df["RETAIL TRANSFERS"].cast("float")
    ).otherwise(None)
)
```

```python
df.printSchema()
```

```python
df.show()
```

### Shows the count, sum, and avg of item retail sales in the Warehouse

```python
df.groupBy("Item Type").count().sort('count', ascending =False).show()
```

```python
df.groupBy("Item Type").sum('RETAIL SALES').sort('sum(RETAIL SALES)', ascending =False).show()
```

```python
df.groupBy("Item Type").avg('RETAIL SALES').sort('avg(RETAIL SALES)', ascending =False).show()
```

### Shows the sum and avg of item warehouse sales

```python
df.groupBy("Item Type").sum('WAREHOUSE SALES').sort('sum(WAREHOUSE SALES)', ascending =False).show()
```

```python
df.groupBy("Item Type").avg('WAREHOUSE SALES').sort('avg(WAREHOUSE SALES)', ascending =False).show()
```
